```python
from google.colab import files

print("📤 Selecionar os 3 arquivos juntos: script (.py), planilha (.xlsx) e dashboard (.html)")
uploaded = files.upload()
print("\nArquivos recebidos:", list(uploaded.keys()))
```

### 3. Organizar os arquivos enviados

Cada arquivo é copiado para um nome fixo, sem espaços ou caracteres especiais — evita erros na etapa seguinte, mesmo que o Colab tenha salvo o upload com um nome diferente do original.

```python
import os, shutil

def find_file(ext):
    matches = [f for f in uploaded if f.lower().endswith(ext)]
    if not matches:
        raise FileNotFoundError(
            f"Nenhum arquivo '{ext}' foi enviado. Recebidos: {list(uploaded.keys())}\n"
            f"Voltar à célula de envio e selecionar os 3 arquivos juntos."
        )
    return matches[0]

def stage_file(ext, target_name):
    """Copia o arquivo enviado para um nome fixo — pula a cópia se já
    estiver com esse nome, evitando erro ao rodar a célula mais de uma vez."""
    source = find_file(ext)
    if os.path.abspath(source) != os.path.abspath(target_name):
        shutil.copy(source, target_name)
    return target_name

script_path = stage_file('.py', 'update_dashboard.py')
xlsx_path   = stage_file('.xlsx', 'planilha.xlsx')
html_path   = stage_file('.html', 'template.html')

print(f"Script:   {script_path}")
print(f"Planilha: {xlsx_path}")
print(f"Template: {html_path}")
```

### 4. Rodar a atualização

O script:
- Localiza cada linha da planilha pelo texto do rótulo (ex.: "Lucro Líquido"), não pelo número da linha
- Extrai a série completa, de 2023 até o trimestre mais recente
- Roda uma validação de sanidade (receita e lucro dentro de um intervalo plausível) antes de publicar
- Gera o JSON com os dados e injeta no HTML

```python
out_name = "porto-dashboard-atualizado.html"

!python {script_path} {xlsx_path} --template {html_path} --out {out_name} --json-out data/dashboard.json
```

### 5. Baixar o dashboard atualizado

```python
from google.colab import files as files_out

files_out.download(out_name)
print(f"✅ Concluído — {out_name} pronto, já com o trimestre mais recente carregado.")
```
